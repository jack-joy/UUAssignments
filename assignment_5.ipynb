**1.** Let's review some basic matrix multiplication. When you have an $M \times N$ matrix $A$ with $M$ rows and $N$ columns, 
$$
A= \left[ \begin{array}{cccc} a_{11} & a_{12} & ... & a_{1N} \\
a_{21} & a_{22} & ... & a_{2N} \\
\vdots & \vdots & ... & \vdots \\
a_{M1} & a_{M2} & ... & a_{MN} 
\end{array} \right],
$$
and you right-multiply it by a vector
$$
x = \left[ \begin{array}{c} x_1 \\ x_2 \\ \vdots \\ x_N 
\end{array} \right],
$$
you get
$$
Ax = \left[ \begin{array}{c} \sum_{i=1}^N a_{1i} x_i \\ \sum_{i=1}^N a_{2i} x_i \\ \vdots \\ \sum_{i=1}^N a_{Mi} x_i 
\end{array} \right].
$$
This is just "matrix row times column vector" element-by-element, stacking the results into a new vector.

For this to make sense, $N$ must be the same for the matrix and the vector, but $M$ can be different from $N$.

$$

```python
taxi = pd.read_pickle('taxicab.pkl')
```

```python
all_neigh = set()
for trip in taxi:
    all_neigh.update(trip.unique())

print(sorted(all_neigh))
print(f"Number of unique Neighborhoods: {len(all_neigh)}")
```

```python
transitions = []

for trip in taxi:
    if len(trip) < 2:
        continue
    for i in range(len(trip) - 1):
        origin = trip.iloc[i]
        dest = trip.iloc[i + 1] 
        transitions.append((origin, dest))

df_trans = pd.DataFrame(transitions, columns=["from", "to"])
transition_counts = pd.crosstab(df_trans["from"], df_trans["to"])
transition_probs = transition_counts.div(transition_counts.sum(axis=1), axis=0)
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(transition_probs, cmap="YlGnBu")
plt.title("Taxicab Neighborhood Transition Matrix (Manhattan Only)")
plt.xlabel("Destination")
plt.ylabel("Origin")
plt.show()
```

Taxi cabs are most likely to order 1, and not 2 or more because taxi cab rides are short and direct for that specific trip. They do not determine the next ride based on past rides. The new rider provides the next destination.

```python
states = sorted(set(transition_probs.index) & set(transition_probs.columns))
transition_probs = transition_probs.loc[states, states]         

start = pd.Series(0.0, index=states)
start["Hell's Kitchen"] = 1.0

def k_step_distribution(start_dist, transition_probs, k):
    # fast + numerically stable
    M = np.linalg.matrix_power(transition_probs.values, k)
    return pd.Series(start_dist.values @ M, index=transition_probs.columns)

for k in [2, 3, 5, 10]:
    dist_k = k_step_distribution(start, transition_probs, k)
    print(f"\nTop destinations after {k} trips from Hell's Kitchen:")
    print(dist_k.sort_values(ascending=False).head(10).round(4))
```

```python
transition_probs = transition_probs.div(transition_probs.sum(axis=1), axis=0).fillna(0)

pi = np.ones(len(transition_probs)) / len(transition_probs)
pi = pd.Series(pi, index=transition_probs.index)
tol = 1e-8  # convergence tolerance
delta = 1
iterations = 0

while delta > tol:
    new_pi = pi @ transition_probs
    delta = np.linalg.norm(new_pi - pi)
    pi = new_pi
    iterations += 1

print(f"Converged after {iterations} iterations.")
steady_state = pi / pi.sum()
steady_state = steady_state.sort_values(ascending=False)

print("\nSteady-state (long-run) distribution:")
print(steady_state.head(10).round(4))
```
